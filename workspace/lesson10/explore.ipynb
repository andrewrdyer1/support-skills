```python
import csv

with open("../data/harbor_invoices.csv") as f:
    rows = list(csv.DictReader(f))

len(rows)
```

The last line of a cell is shown automatically, so you don't need `print` to look at something. Look at one row:

```python
rows[0]
```

Count invoices by status. `Counter` does the loop-and-`.get()` pattern from Lesson 8 for you:

```python
from collections import Counter

Counter(row["Status"] for row in rows)
```

The overdue ones, as (number, client, amount):

```python
overdue = [row for row in rows if row["Status"] == "overdue"]
[(row["Number"], row["Client"], row["Amount"]) for row in overdue]
```
